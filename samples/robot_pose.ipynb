```python
# 質量中心の位置
com = robot.centerOfMass
print(f"Center of Mass:  {com}")

# ルートリンクの位置
root_pos = robot.rootCoords().pos
print(f"Root position: {root_pos}")

# 各リンクの位置から範囲を計算
link_positions = [link.p for link in robot.linkList]
x_coords = [p[0] for p in link_positions]
y_coords = [p[1] for p in link_positions]
z_coords = [p[2] for p in link_positions]

print(f"\nRobot dimensions:")
print(f"X range: {min(x_coords):.3f} to {max(x_coords):.3f} (width:  {max(x_coords) - min(x_coords):.3f} m)")
print(f"Y range: {min(y_coords):.3f} to {max(y_coords):.3f} (depth: {max(y_coords) - min(y_coords):.3f} m)")
print(f"Z range:  {min(z_coords):.3f} to {max(z_coords):.3f} (height: {max(z_coords) - min(z_coords):.3f} m)")

print(f"\nTotal height: {max(z_coords) - min(z_coords):.3f} m")
```
